# Projectile Motion
Calculates range, maximum height, and trajectory of a projectile.

```python
import math
import numpy as np

g = 9.81
v_0 = 25.0
angle_deg = 45.0
theta = math.radians(angle_deg)
```

## Flight Time and Range

```python
t_flight = 2 * v_0 * math.sin(theta) / g
max_range = (v_0 ** 2) * math.sin(2 * theta) / g
print(f"Flight time: {t_flight:.2f} s")
print(f"Max range:   {max_range:.2f} m")
```

## Trajectory Plot

```python
import matplotlib.pyplot as plt

t = np.linspace(0, t_flight, 200)
x = v_0 * math.cos(theta) * t
y = v_0 * math.sin(theta) * t - 0.5 * g * (t ** 2)

plt.figure(figsize=(6.5, 3.2))
plt.plot(x, y, label="y(x)")
plt.xlabel("x [m]")
plt.ylabel("y [m]")
plt.title("Projectile Trajectory")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
